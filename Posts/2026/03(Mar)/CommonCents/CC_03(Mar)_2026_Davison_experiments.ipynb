```python
import matplotlib        as mpl
import matplotlib.pyplot as plt
import numpy             as np
import pickle
```

```python
def single_bet(x):
    if x <= 1/52:
        return -4900.0
    else:
        return 100.0
```

```python
# Load the data from the pickle file
with open('CC_03(Mar)_2026_Q2_bet.pickle', 'rb') as file:
    winnings = pickle.load(file)
```

```python
num_trials = 16
fig, axes = plt.subplots(4,4,figsize=(8,8))
axes = axes.flatten()
for i in range(num_trials):
    axes[i].plot(np.cumsum(winnings[i])/1000)
    axes[i].set_ylim([-50,50])
    axes[i].axhline(y=0, color='red', linestyle='-', linewidth=2)
    if i % 4 != 0:
        axes[i].tick_params(axis='both', which='both',left=False, labelleft=False)
    if i % 4 == 0:
        axes[i].set_ylabel('Winnings ($k)')
    if i < 12:
        axes[i].tick_params(axis='both', which='both',bottom=False,labelbottom=False)
    if i >= 12:
        axes[i].set_xlabel('Trial')
```
