```python
import sympy as sym
```

```python
r1, r2, x, y, z = sym.symbols('r_1 r_2 x y z',real=True)
mu = sym.symbols('mu',real=True,positive=True)
d1 = sym.sqrt( (x+mu)**2   + y**2 + z**2 )
d2 = sym.sqrt( (x-1+mu)**2 + y**2 + z**2 )
U  = sym.Rational(1,2)*(x**2 + y**2) + (1-mu)/d1 + mu/d2
A  = sym.symbols('A',real=True,positive=True)
q  = sym.symbols('q')
display(U)
```

```python
coords = [x,y,z]
```

```python
dU = {}
for coord in coords:
    dU[(coord,)] = sym.diff(U,coord)
display(dU)
```

```python
ddU = {}
for coord1 in coords:
    for coord2 in coords:
        ddU[(coord1,coord2)] = sym.diff(dU[(coord1,)],coord2)

for ddUk in ddU:
    display(ddUk,sym.latex(ddU[ddUk]))
```

```python
colinear_subs       = {y:0,z:0}
equilateral_subs_L4 = {x:sym.Rational(1,2)-mu,y:sym.sqrt(3)/2,z:0}
equilateral_subs_L5 = {x:sym.Rational(1,2)-mu,y:-sym.sqrt(3)/2,z:0}
```

```python
dU_colinear   = {}
for k in dU.keys():
    dU_colinear[k] = dU[k].subs(colinear_subs)

dU_L4 = {}
for k in dU.keys():
    dU_L4[k] = sym.simplify(dU[k].subs(equilateral_subs_L4))

dU_L5 = {}
for k in dU.keys():
    dU_L5[k] = sym.simplify(dU[k].subs(equilateral_subs_L5))

display(dU_colinear)
display(dU_L4)
display(dU_L5)
display(dU_colinear[(x,)])
```

## Colinear Points

```python
ddU_colinear  = {}
for k in ddU.keys():
    ddU_colinear[k] = sym.simplify(ddU[k].subs(colinear_subs))
    ddU_colinear[k] = sym.simplify(ddU_colinear[k].subs({d1.subs(colinear_subs):r1,d2.subs(colinear_subs):r2}))
    ddU_colinear[k] = sym.simplify(ddU_colinear[k].subs({d1.subs(colinear_subs)**2:r1**2,d2.subs(colinear_subs)**2:r2**2})).collect(r1)
```

```python
ddU_colinear[(x,x)].collect(r1)
```

```python
for ddUc2 in ddU_colinear:
    display(ddUc2,ddU_colinear[ddUc2],sym.latex(ddU_colinear[ddUc2]))
```

```python
M = sym.Matrix([[0,0,1,0],[0,0,0,1],[1+2*A,0,0,2],[0,1-A,-2,0]])
colinear_charpoly = M.charpoly(q)
display(colinear_charpoly)
for sol in sym.solve(colinear_charpoly,q**2):
    display(sol)
```

## $L_4$

```python
ddU_L4 = {}
for k in ddU.keys():
    ddU_L4[k] = sym.simplify(ddU[k].subs(equilateral_subs_L4))
    ddU_L4[k] = sym.simplify(ddU_L4[k].subs({d1.subs(equilateral_subs_L4):r1,d2.subs(equilateral_subs_L4):r2}))
    ddU_L4[k] = sym.simplify(ddU_L4[k].subs({d1.subs(equilateral_subs_L4)**2:r1**2,d2.subs(equilateral_subs_L4)**2:r2**2}))
    ddU_L4[k] = sym.simplify(ddU_L4[k].subs({r1:1,r2:1}))
```

```python
M = sym.Matrix([[0,0,1,0],[0,0,0,1],[sym.Rational(3,4),(1-2*mu)*3*sym.sqrt(3)/4,0,2],[(1-2*mu)*3*sym.sqrt(3)/4,sym.Rational(9,4),-2,0]])
M.charpoly(q)
```

## $L_5$

```python
ddU_L5 = {}
for k in ddU.keys():
    ddU_L5[k] = sym.simplify(ddU[k].subs(equilateral_subs_L5))
    ddU_L5[k] = sym.simplify(ddU_L5[k].subs({d1.subs(equilateral_subs_L5):r1,d2.subs(equilateral_subs_L5):r2}))
    ddU_L5[k] = sym.simplify(ddU_L5[k].subs({d1.subs(equilateral_subs_L5)**2:r1**2,d2.subs(equilateral_subs_L5)**2:r2**2}))
    ddU_L5[k] = sym.simplify(ddU_L5[k].subs({r1:1,r2:1}))
```
