```python
import matplotlib        as mpl
import matplotlib.pyplot as plt
import numpy             as np
```

```python
mu = 3.04e-6
```

```python
dx = 0.00001
x  = np.arange(0,0.02,dx)
X  = 0.01
Y  = 0
```

```python
y = x**5 + (3-mu)*x**4 + (3-2*mu)*x**3 - mu*x**2 - 2*mu*x - mu
```

```python
fig, ax = plt.subplots(1,1,figsize=(10,5))
ax.plot(x,y)
ax.plot(X,Y,'bo')
ax.grid('on')
ax.set_ylim([-1e-5,1e-5])
ax.set_xlabel(r'$\gamma$')
ax.set_ylabel(r'Master Equation - Region II')
fig.savefig('Region_II_master_equation_Sun-Earth-Moon-L2.png')
```
